```python
from ising_genetic_new import genetic_algorithm
from ising import monte_carlo_metropolis
from matplotlib import pyplot as plt
from timeit import timeit
import numpy as np
```

```python
betac = np.log(1.0 + np.sqrt(2)) / 2.0
```

```python
times = [
    timeit(
        f"monte_carlo_metropolis(n=60, beta=0.01 * {betac}, h=0.0, max_steps=1_000, initial_state=1, method='checkerboard')",
        number=10,
        setup="from ising import monte_carlo_metropolis",
    ),
    timeit(
        f"genetic_algorithm(60, 0.01 * {betac}, 0.0, n_states=100, n_children=10, n_cycles=1_000, mutation_rate=0.01)",
        number=10,
        setup="from ising_genetic import genetic_algorithm",
    ),
]
```

```python
fig, ax = plt.subplots()

ax.bar(["Monte Carlo", "Genetic algorithm"], times)
ax.set_ylabel("Time (s)")
fig.suptitle("Comparison of average run time over 10 runs")
```
